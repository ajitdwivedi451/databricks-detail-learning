### View & Temp View

```sql
Create OR REPLACE VIEW dev.bronze.client_vw 
As
Select * from dev.bronze.client where id in (1,2)
```

```sql
Select * from dev.bronze.client_vw
```

```sql
Create Temp view client1_vw
as
Select * from dev.bronze.client where id in (1,2)
```

```sql
Select * from client1_vw
```

- ### Shallow clone and deep Clone

```sql
Create table dev.bronze.client_dclone DEEP CLONE dev.bronze.client
```

```sql
describe history dev.bronze.client_dclone
```

```sql
describe extended dev.bronze.client_dclone
```

```sql
Create table dev.bronze.client_sclone SHALLOW CLONE dev.bronze.client
```

```sql
describe extended dev.bronze.client_sclone
```

```sql
describe history dev.bronze.client_sclone
```

```sql
SELECT * from dev.bronze.client_sclone
```
